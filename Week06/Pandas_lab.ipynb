```python
# B3 — STANDARDIZE TEXT

# B3.1. Convert program and city to Pandas string types and remove leading and trailing spaces.

students["program"] = (students["program"].astype("string").str.strip())

students["city"] = (students["city"].astype("string").str.strip())

students[["program", "city"]]
```

#### Part C — Types, ranges, and missing values (35 minutes)

##### C1. Convert types

```python
assessment_columns = ["attendance", "assignment", "midterm", "final"]
students[assessment_columns].dtypes
```

```python
# C1.1. Convert attendance, assignment, midterm, and final to numeric with errors="coerce".
assessment_columns = ["attendance", "assignment", "midterm", "final"]
conversion_columns = assessment_columns + ["admission_date"]
missing_before_conversion = students[conversion_columns].isna().sum().copy()
original_dates = students["admission_date"].copy()
for column in assessment_columns:
    students[column] = pd.to_numeric(students[column], errors="coerce")
students[assessment_columns].dtypes
```

```python
   # C1.2. Parse mixed admission dates with dayfirst=True and errors="coerce".
   # ISO-style dates (2025-09-01, 2025/09/01) year-first hain, unko dayfirst se nahi parse karna.
   s = students["admission_date"].astype("string")
   year_first = s.str.match(r"^\d{4}[-/]")

   parsed_iso = pd.to_datetime(s.where(year_first), format="mixed", errors="coerce")
   parsed_dayfirst = pd.to_datetime(s.where(~year_first), format="mixed", dayfirst=True, errors="coerce")

   students["admission_date"] = parsed_iso.fillna(parsed_dayfirst)
   students["admission_date"]
```

```python
# C1.3. Convert student_id to string and display dtypes and newly introduced missing counts.
students["student_id"] = students["student_id"].astype("string")
new_missing = students[conversion_columns].isna().sum() - missing_before_conversion
print(students.dtypes)
print("New missing values:", new_missing, sep="\n")
```

##### C2. Validate ranges

```python
# C2.1. Identify values outside 0–100 and store affected records for review.
invalid_mask = (students[assessment_columns] < 0) | (students[assessment_columns] > 100)
invalid_records = students.loc[invalid_mask.any(axis=1)].copy()
invalid_records
```

```python
# C2.2. Record each invalid value, the cleaning decision, and the reason in cleaning_log.
for column in assessment_columns:
    for _, row in students.loc[invalid_mask[column]].iterrows():
        cleaning_log.append({
            "student_id": row["student_id"], "column": column,
            "original_value": row[column], "decision": "Replace with missing value",
            "reason": "Outside the valid range 0–100"
        })
pd.DataFrame(cleaning_log)
```

```python
# C2.3. Replace invalid values with pd.NA using .loc and verify no out-of-range values remain.
for column in assessment_columns:
    students.loc[invalid_mask[column], column] = pd.NA

remaining_invalid = (students[assessment_columns] < 0) | (students[assessment_columns] > 100)
assert not remaining_invalid.any().any()
print("Remaining invalid values:", int(remaining_invalid.sum().sum()))
```

##### C3. Handle missing values

```python
# C3.1. Record missing counts and affected IDs before imputation; retain flags for imputed values.
missing_before_imputation = students[original_columns].isna().sum().copy()
affected_ids = {}
for column in assessment_columns:
    affected_ids[column] = students.loc[students[column].isna(), "student_id"].tolist()
    students[f"{column}_imputed"] = students[column].isna()
print("Affected IDs:", affected_ids)
missing_before_imputation
```

```python
# C3.2. Fill attendance with its program median, then use the original overall median as fallback.
overall_median = students["attendance"].median()
program_median = students.groupby("program")["attendance"].transform("median")
attendance_fill = program_median.fillna(overall_median)

for index in students.index[students["attendance_imputed"]]:
    cleaning_log.append({
        "student_id": students.loc[index, "student_id"], "column": "attendance",
        "original_value": "missing", "decision": f"Fill with {attendance_fill.loc[index]}",
        "reason": "Program median; original overall median if group has no valid values"
    })

students["attendance"] = students["attendance"].fillna(attendance_fill)
students[["student_id", "program", "attendance", "attendance_imputed"]]
```

```python
# C3.3. Fill missing assessment marks with column medians and log the rule.
for column in ["assignment", "midterm", "final"]:
    median_value = students[column].median()
    for student_id in affected_ids[column]:
        cleaning_log.append({
            "student_id": student_id, "column": column, "original_value": "missing",
            "decision": f"Fill with column median {median_value}",
            "reason": "Classroom simulation only; retained imputation flag"
        })
    students[column] = students[column].fillna(median_value)

students[["student_id", "assignment", "midterm", "final",
          "assignment_imputed", "midterm_imputed", "final_imputed"]]
```

```python
# C3.4. Store missing_after, compare counts, and verify scoring columns have no missing values; state limitations.
missing_after = students[original_columns].isna().sum()
assert students[assessment_columns].notna().all().all()

pd.DataFrame({"Before imputation": missing_before_imputation, "After": missing_after})
```

#### Part D — Create features and summarize

##### D1. Create features

```python
# D1.1. Calculate weighted_score using assignment 20%, midterm 30%, and final 50% for this exercise.
students["weighted_score"] = (
    students["assignment"] * 0.20 + students["midterm"] * 0.30 + students["final"] * 0.50
)
# These are exercise weights, not the official course assessment policy.
students[["student_id", "weighted_score"]]
```

```python
# D1.2. Create attendance_status: Low below 75, Satisfactory from 75 to below 85, Excellent from 85.
import numpy as np

students["attendance_status"] = np.select(
    [students["attendance"] >= 85, students["attendance"] >= 75],
    ["Excellent", "Satisfactory"], default="Low"
)
students[["student_id", "attendance", "attendance_status"]]
```

```python
# D1.3. Create grades A >= 85, B >= 75, C >= 65, D >= 50, otherwise F, using np.select().
conditions = [students["weighted_score"] >= 85,
              students["weighted_score"] >= 75,
              students["weighted_score"] >= 65,
              students["weighted_score"] >= 50]
students["grade"] = np.select(conditions, ["A", "B", "C", "D"], default="F")
students[["student_id", "weighted_score", "grade"]]
```

```python
# D1.4. Flag at_risk when attendance < 75 or weighted_score < 50; display features and verify score bounds.
students["at_risk"] = (students["attendance"] < 75) | (students["weighted_score"] < 50)
assert students["weighted_score"].between(0, 100).all()
students[["student_id", "weighted_score", "attendance_status", "grade", "at_risk"]]
```

##### D2. Group and summarize

```python
# D2.1. Create program_summary with student count and average weighted score for each program.
program_summary = students.groupby("program").agg(
    student_count=("student_id", "count"), average_score=("weighted_score", "mean")
).reset_index()
program_summary
```

```python
# D2.2. Create and display grade_distribution, then verify both summary counts equal the student count.
grade_distribution = (students["grade"].value_counts().sort_index()
    .rename_axis("grade").reset_index(name="count"))
assert program_summary["student_count"].sum() == len(students)
assert grade_distribution["count"].sum() == len(students)
grade_distribution
```
