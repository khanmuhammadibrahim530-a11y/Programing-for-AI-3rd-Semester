```python
import pandas as pd

students = pd.read_csv(
 "data/raw/students_messy.csv",
 na_values=["", "missing", "N/A", "-"]
)
# print 
print("--HEAD--")
print(students.head())
print("--TAIL--")
print(students.tail())
print("--INFO--")
print(students.info())
print("--DESCRIPTION--")
print(students.describe())
print("--COLUMNS--")
print(students.columns)
print("--INDEX--")
print(students.index)

print("--SHAPE--")
print(students.shape)

print("--DTYPES--")
print(students.dtypes)
print("--MISSING VALUES--")
print(students.isna().sum())
print("--DUPLICATES--")
print("Exact Duplicates:",students.duplicated().sum())

print("Duplicate Rows:",students[students.duplicated()])
print("--UNIQUE VALUES--")
print("Missing Values:", students.isnull().sum().sum())
print("Program labels:")
print(students['program'].unique())
```
