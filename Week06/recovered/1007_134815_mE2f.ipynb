```python
import pandas as pd

students = pd.read_csv(
 "data/raw/students_messy.csv",
 na_values=["", "missing", "N/A", "-"]
)
print(students.head())
print(students.shape)
print(students.dtypes)
print(students.isna().sum())
print("Exact duplicates:", students.duplicated().sum())
```

#### B1 Select and filter


```python
# Remove extra spaces from text columns
for col in students.select_dtypes(include=["object"]).columns:
    students[col] = students[col].astype(str).str.strip()


# 1. Performance View
perf_cols = ["student_id", "name", "attendance", "assignment", "midterm", "final"]
print("\nPerformance View:\n", students[perf_cols])


# 2. AI Students
ai_students = students[
    students["program"].str.contains("AI", case=False, na=False)
]

print("\nAI Students:")
print(ai_students[["student_id", "name", "program"]].to_string(index=False))


# 3. Students below 75% attendance
print("\nStudents below 75% attendance:\n",
      students[students["attendance"] < 75])


print("\nStudents with final marks at least 80:\n",
      students[students["final"] >= 80])


# 5. Sort by final descending, missing values last
print("\nSorted by final:\n",
      students.sort_values("final", ascending=False, na_position="last"))
```
