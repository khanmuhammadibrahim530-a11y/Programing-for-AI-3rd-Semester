```python
from pathlib import Path

import pandas as pd 

DATA_PATH = Path("F:/BS AI 3rd Semester/Programingforai/Programing-for-AI-2026/Week06/data/processed/students_clean.csv")
FIGURE_DIR = Path("F:/BS AI 3rd Semester/Programingforai/Programing-for-AI-2026/Week7/week07/figures") 
FIGURE_DIR.mkdir(parents=True, exist_ok=True)  
students = pd.read_csv(DATA_PATH) 
required = {"program", "attendance", "final", "weighted_score", "grade"}  
print(students.shape) 
print(students[sorted(required)].head())
print(students[sorted(required)].isna().sum()) 
```


# Line plot -ordered trend


```python
import matplotlib.pyplot as plt

weeks = [1, 2, 3, 4, 5, 6]
average_attendance = [78, 80, 81, 79, 83, 85]
fig, ax = plt.subplots(figsize=(8, 5)) 
ax.plot(weeks, average_attendance, marker="o", color="#EB2525")
ax.set(title="Weekly Average Attendance", xlabel="Week", ylabel="Attendance (%)")
ax.set_xticks(weeks) 
ax.grid(axis="y", alpha=0.3) 
fig.tight_layout() 
```

# B2.Scatter plot: relationship 


```python
# 1. Setup the plot
fig, ax = plt.subplots(figsize=(7, 4))

# 2. Create the scatter plot
ax.scatter(students["attendance"], students["final"], color="#EB2525", alpha=0.7)

# 3. Add title, labels, and grid
ax.set(title="Attendance vs Final Marks",
       xlabel="Attendance (%)", ylabel="Final Marks")
ax.grid(alpha=0.3)

# 4. Adjust layout
fig.tight_layout()

# 5. Show the plot
plt.show()
```

C1 — Histogram

```python
# 1. Setup the plot
fig, ax = plt.subplots(figsize=(7, 4))

# 2. Create the histogram
ax.hist(students["weighted_score"].dropna(), bins=10,
        color="#EB2525", edgecolor="white")

# 3. Add title, labels, and grid
ax.set(title="Distribution of Weighted Scores",
       xlabel="Weighted Score", ylabel="Number of Students")
ax.grid(axis="y", alpha=0.3)

# 4. Adjust layout
fig.tight_layout()

# 5. Show the plot
plt.show()
```

#C2 — Box Plot

```python
# 1. Setup the plot
fig, ax = plt.subplots(figsize=(7, 4))

# 2. Create the box plot
ax.boxplot(students["weighted_score"].dropna(), vert=False)

# 3. Add title, label, and grid
ax.set(title="Weighted Score Summary", xlabel="Weighted Score")
ax.grid(axis="x", alpha=0.3)

# 4. Adjust layout
fig.tight_layout()

# 5. Show the plot
plt.show()
```

#C3 — Bar Chart

```python
# 1. Calculate average score by program
program_scores = students.groupby("program")["weighted_score"].mean().sort_values()

# 2. Setup the plot
fig, ax = plt.subplots(figsize=(7, 4))

# 3. Create the bar chart
ax.bar(program_scores.index, program_scores.values, color="#EB2525")

# 4. Add title, labels, and grid
ax.set(title="Average Weighted Score by Program",
       xlabel="Program", ylabel="Average Weighted Score")
ax.tick_params(axis="x", rotation=25)
ax.grid(axis="y", alpha=0.3)

# 5. Adjust layout
fig.tight_layout()

# 6. Show the plot
plt.show()
```

#D1 — Student Performance Dashboard

```python
# 1. Setup the dashboard
fig, axes = plt.subplots(2, 2, figsize=(11, 8))

# 2. Create histogram
axes[0, 0].hist(students["weighted_score"].dropna(), bins=10,
                color="#EB2525", edgecolor="white")
axes[0, 0].set(title="Score Distribution",
               xlabel="Score", ylabel="Count")

# 3. Create scatter plot
axes[0, 1].scatter(students["attendance"], students["final"],
                   color="#EB2525", alpha=0.7)
axes[0, 1].set(title="Attendance vs Final",
               xlabel="Attendance (%)", ylabel="Final Marks")

# 4. Create bar chart
axes[1, 0].bar(program_scores.index, program_scores.values,
               color="#EB2525")
axes[1, 0].set(title="Average Score by Program",
               xlabel="Program", ylabel="Average Score")
axes[1, 0].tick_params(axis="x", rotation=25)

# 5. Create box plot
axes[1, 1].boxplot(students["weighted_score"].dropna(), vert=False)
axes[1, 1].set(title="Weighted Score Summary",
               xlabel="Score")

# 6. Add grid to all charts
for ax in axes.flat:
    ax.grid(alpha=0.2)

# 7. Add dashboard title
fig.suptitle("Student Performance Dashboard")

# 8. Adjust layout
fig.tight_layout()

# 9. Show the dashboard
plt.show()
```

#D2 — Dashboard Save Karna

```python
# 1. Set file paths
png_path = FIGURE_DIR / "student_dashboard.png"
pdf_path = FIGURE_DIR / "student_dashboard.pdf"

# 2. Save dashboard
fig.savefig(png_path, dpi=300, bbox_inches="tight")
fig.savefig(pdf_path, bbox_inches="tight")

# 3. Check saved files
print(png_path.exists())
print(pdf_path.exists())

# 4. Show dashboard
plt.show()

# 5. Close the figure
plt.close(fig)
```
